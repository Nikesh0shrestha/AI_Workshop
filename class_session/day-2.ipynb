```python

import pandas as pd
import numpy as np
# creating a DataFrame from  dictionary

data = {
    'student_name' : ['Alice','Bob','Charlie','Diana','Eve','Alice'],
    'hours_studied' : [65,8,3,10,6,65],
    'pervious_score' : [65,85,55,92,75,65],
    'attendance' : [80,90,65,95,85,80],
    'sleep_housrs' : [7.5,8.0,6.0,9.0,7.0,7.5],
    'extracurricular' : ['Yes','No','Yes','Yes','No','Yes'],
    'parent_education' : ['Bachelor','Master','High School','PHD','Bachelor','Bachelor']
}

df = pd.DataFrame(data)
print("Student Performance DataFrame:")
print(df)

print("Data Info:")
print(df.info())


#  Basic DataFrame information
print("\nDataFrame shape:",df.shape[0],"rows,",df.shape[1],"columns")
print("Column nanes:",df.columns.tolist())
print("Data types:\n", df.dtypes)


```

```python
# view data 
print("First 5 rows:")
print(df.head(5))

# view last 2 rows
print("\n Last 2 rows:")
print(df.tail(2))

# Statistical summary
print("\n Statistical Summary:")
print(df.describe())

# Check for missing values
print("\n Missing values:")
print(df.isnull().sum())

# check for nan
print("\nCheck for nan Values:")
print(df.isna().sum()) 

# check for duplicates
print("\n Duplicates Rows:")
print(df.duplicated().sum())

# Remove the dulicates 
print("Removing the duplicates:")
print(df.drop_duplicates)


```

```python
def calculate_grade(score):
    if score >= 90:
        return "A"
    elif score >= 80:
        return 'B'
    elif score >= 70:
        return 'C'
    elif score >= 60:
        return 'D'
    else:
        return 'F'

df['grade'] =  df['pervious_score'].apply(calculate_grade)
print("DataFrame with new columns:,", df)

# using map for categorical data.
extracurricular_map = {'Yes': 1, 'No': 0}
df['extracurricular_binary'] = df['extracurricular'].map(extracurricular_map)

```
